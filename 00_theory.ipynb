### PySpark & Jupyter Notebooks: Theoretical Foundation

#### 1. Core Architecture of PySpark
PySpark is the Python API for Apache Spark, an open-source, distributed cluster-computing framework designed for fast big-data processing.

* **Driver Program:** The process running the `SparkSession` that creates the SparkContext. It coordinates workers, converts user code into tasks, and schedules them.
* **Executor:** Worker nodes running on the cluster (or locally) that execute the tasks assigned by the driver and store data in memory or disk.
* **SparkSession (`spark`):** The unified entry point to programming Spark with the Dataset and DataFrame API, combining `SparkContext`, `SQLContext`, and `HiveContext`.

#### 2. Key Data Abstractions
* **RDD (Resilient Distributed Dataset):** Spark's fundamental data structure. It is an immutable, lazy-evaluated, fault-tolerant collection of records partitioned across cluster nodes.
* **DataFrame:** Built on top of RDDs, DataFrames are organized into named columns (like a relational database table). They feature automated query optimization via the **Catalyst Optimizer**.
* **Dataset:** An extension of DataFrames providing compile-time type safety (primarily used in Scala/Java; PySpark DataFrames are conceptually Datasets of `Row`).

#### 3. Transformations vs. Actions
Spark operations are divided into two categories to optimize performance through lazy evaluation:

* **Transformations (Lazy):** Operations (e.g., `select()`, `filter()`, `groupBy()`, `join()`) that do not compute results immediately. Instead, they build up a logical execution plan (lineage graph).
* **Actions (Eager):** Operations (e.g., `show()`, `collect()`, `count()`, `write()`) that trigger the execution of the lineage graph, returning results to the driver or persisting them to storage.
