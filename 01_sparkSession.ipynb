### Creating SparkSession for our code

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
    .builder
    .appName("Spark Introduction")
    .getOrCreate()
)
```

```python
spark
```

```python
## data and schema

from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType, DateType
from pyspark.sql import Row
from datetime import date

# Define the employee schema
emp_schema = StructType([
    StructField("emp_id", IntegerType(), False),
    StructField("emp_name", StringType(), True),
    StructField("department", StringType(), True),
    StructField("salary", DoubleType(), True),
    StructField("hire_date", DateType(), True),
    StructField("country", StringType(), True),
])

# 20 rows of sample employee data
emp_rows = [
    Row(1, "John Smith", "Engineering", 85000.0, date(2021, 3, 15), "USA"),
    Row(2, "Alice Johnson", "Engineering", 92000.0, date(2020, 7, 1), "USA"),
    Row(3, "Bob Williams", "Sales", 65000.0, date(2022, 1, 20), "UK"),
    Row(4, "Carol Brown", "Sales", 68000.0, date(2019, 11, 5), "UK"),
    Row(5, "David Jones", "HR", 55000.0, date(2021, 9, 10), "Canada"),
    Row(6, "Eva Garcia", "HR", 58000.0, date(2023, 2, 28), "Canada"),
    Row(7, "Frank Miller", "Engineering", 78000.0, date(2018, 5, 22), "Germany"),
    Row(8, "Grace Davis", "Marketing", 72000.0, date(2022, 8, 14), "Germany"),
    Row(9, "Henry Wilson", "Marketing", 70000.0, date(2020, 12, 1), "India"),
    Row(10, "Ivy Taylor", "Finance", 95000.0, date(2017, 6, 30), "India"),
    Row(11, "Jack Anderson", "Finance", 89000.0, date(2021, 4, 18), "Australia"),
    Row(12, "Karen Thomas", "Engineering", 81000.0, date(2023, 3, 7), "Australia"),
    Row(13, "Leo Moore", "Sales", 62000.0, date(2019, 10, 25), "USA"),
    Row(14, "Mia Martin", "Engineering", 90000.0, date(2022, 5, 11), "USA"),
    Row(15, "Nick Lee", "HR", 60000.0, date(2018, 2, 9), "UK"),
    Row(16, "Olivia Harris", "Marketing", 74000.0, date(2020, 9, 3), "Canada"),
    Row(17, "Paul Clark", "Finance", 87000.0, date(2023, 7, 19), "Germany"),
    Row(18, "Quinn Lewis", "Engineering", 83000.0, date(2021, 12, 12), "India"),
    Row(19, "Rachel Walker", "Sales", 66000.0, date(2019, 8, 8), "Australia"),
    Row(20, "Sam Hall", "Marketing", 71000.0, date(2022, 11, 27), "USA"),
]

# Create the DataFrame
emp_df = spark.\
    createDataFrame(emp_rows, schema=emp_schema)

emp_df.printSchema()
emp_df.show(20, truncate=False)
```

```python
# number of partittion
#emp_df.rdd.getNumPartitions()
```

```python
# applying our first transformation (salary > 72k)

emp_salary_greater_72k = emp_df.\
                                where("salary > 72000")

emp_salary_greater_72k.show()
```

```python
from pyspark.sql.functions import spark_partition_id

# number of partitions
emp_df.select(spark_partition_id().alias("partition_id")).distinct().count()
```

```python
# Write the employee DataFrame to CSV format
emp_df.write\
    .mode("overwrite")\
    .option("header", True)\
    .csv("/Volumes/pyspark/data/pyspark/emp_csv")
```
