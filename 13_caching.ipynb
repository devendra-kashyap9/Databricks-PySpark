```python
from pyspark.sql import SparkSession

spark = SparkSession.\
            builder.\
                appName('Spark Data Persistence').\
                    getOrCreate()
```

```python
## reading a csv file
csv_path = "/Volumes/pyspark/data/pyspark/employee.csv"
_schema = {
    'header': True,
    'inderschema': True
}

df = spark.read.format("csv").\
            options(**_schema).\
                load(csv_path)
```

```python
df.where('salary > 65000').show()
```

```python
# Cache DataFrame (Cache or Persist)
df.cache().count()    # Default storage is Disk Memory & Deserialized but for rdd it is Memory ONLY
```

```python
# this time it will be quicker
df.where('salary > 65000').show()
```

#### NOTE: Whenever we have created `cache` spark will run it through cache rather thna the origianl file; If we `create cache on partial data` it will look for the original file or vice-versa depending on the conditions on which cache has been created.

```python
# Memory mode in pyspark
import pyspark

df_persist = df.persist(pyspark.storagelevel.StorageLevel.MEMORY_ONLY)

df_persist.write.format('noop').mode('overwrite').save()
```

```python
df_persist.show()
```

#### Difference Between `cache` and `persist` is their memory options for cache it ie `DISK_MEMORY`and data is `DESERIALIZED` while for later we have provided with plethora of options.

```python
# Remove cache
df.unpersist()
```

```python
# remove cache
spark.catalog.clearCache()
```
