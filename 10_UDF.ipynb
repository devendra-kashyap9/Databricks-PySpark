```python
from pyspark.sql import SparkSession

spark = SparkSession.\
            builder.\
                appName("User Defined Functions (UDF)").\
                    getOrCreate()
```

```python
## reading a csv file
csv_path = "/Volumes/pyspark/data/pyspark/employee.csv"
_schema = {
    'header': True,
    'inderschema': True
}

df = spark.read.format("csv").\
            options(**_schema).\
                load(csv_path)

df.show()
```

```python
# Create a function which will give 10% bonus of salary
def bonus(salary):
    return salary * 0.1
```

```python
# Register our UDF
from pyspark.sql.functions import udf
from pyspark.sql.types import FloatType

bonus_udf = udf(bonus, FloatType())
df = df.withColumn('bonus', bonus_udf(df.salary.cast('float')))
display(df)
```

```python
# to use the udf for SparkSQL we have to register it
from pyspark.sql.functions import expr

spark.udf.register("bonus_sql_udf", bonus, FloatType())
df.withColumn('bonus', expr("bonus_sql_udf(CAST(salary AS FLOAT))")).show()
```
